# Dowload Data

In this section, we will download the **training and test images** required for melanoma classification from GoogleDrive. These datasets will be used to train and evaluate our classification model.


```python
#Librerías 
import os
import zipfile
import requests
import gdown # para descargar archivos desde Google Drive
```

```python
# ID del archivo en Google Drive
file_id = "13aTjjp9R3uOAJRn3zuZIJXLH16gTzTCP"
url = f"https://drive.google.com/uc?id={file_id}"


output_dir = "../data"   
output = os.path.join(output_dir, "images.zip")
os.makedirs(output_dir, exist_ok=True)

gdown.download(url, output, quiet=False)
with zipfile.ZipFile(output, "r") as zip_ref:
    zip_ref.extractall(output_dir)

print(f"✅ Dataset listo en '{output_dir}/images/'")

```
